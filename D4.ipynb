```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=df.drop_duplicates()
```

```python
sns.boxplot(df['Age'])
```

```python
x=df.drop(columns='Heart_Risk')
```

```python
y=df['Heart_Risk']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
SS=StandardScaler()
x_train_scaled=SS.fit_transform(x_train)
x_test_scaled=SS.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
LR=LogisticRegression()
LR.fit(x_train_scaled,y_train)
y_pred=LR.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix,f1_score,classification_report
```

```python
accuracy_score(y_test,y_pred)
```

```python
baseline_accuracy=accuracy_score(y_test,y_pred)
```

```python
print(confusion_matrix(y_test,y_pred))
```

```python
print(classification_report(y_test,y_pred))
```

```python
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    LR,
    x_train_scaled,
    y_train,
    cv=5,
    scoring='accuracy'
)

print("CV Scores:", cv_scores)
print("Mean CV Accuracy:", cv_scores.mean())
```

```python
cv_accuracy=cv_scores.mean()
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression

model_lr = LogisticRegression(max_iter=1000)

classifier = GridSearchCV(
    model_lr,
    {
        'C': [0.01, 0.1, 1, 10, 100]
    },
    cv=5,
    return_train_score=False
)

classifier.fit(x_train_scaled, y_train)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression

model_lr = LogisticRegression(max_iter=1000)

classifier = GridSearchCV(
    model_lr,
    {
        'C': [0.01, 0.1, 1, 10, 100]
    },
    cv=5,
    return_train_score=False
)

classifier.fit(x_train_scaled, y_train)
```

```python
classifier.best_params_
```

```python
classifier.best_score_
```

```python
results_grid = pd.DataFrame(classifier.cv_results_)
results_grid
```

```python
y_pred_grid = classifier.predict(x_test_scaled)
from sklearn.metrics import accuracy_score
grid_accuracy = accuracy_score(y_test, y_pred_grid)
print("GridSearch Accuracy:", grid_accuracy)
```

```python
from sklearn.model_selection import RandomizedSearchCV

classifier_r = RandomizedSearchCV(
    model_lr,
    {
        'C': [0.001, 0.01, 0.1, 1, 10, 100, 1000]
    },
    n_iter=5,
    cv=5,
    return_train_score=False,
    random_state=42
)

classifier_r.fit(x_train_scaled, y_train)
```

```python
classifier_r.best_params_
```

```python
classifier_r.best_score_
```

```python
results_random = pd.DataFrame(classifier_r.cv_results_)
results_random
```

```python
y_pred_random = classifier_r.predict(x_test_scaled)

random_accuracy = accuracy_score(y_test, y_pred_random)

print("RandomizedSearch Accuracy:", random_accuracy)
```

```python
from sklearn.ensemble import RandomForestClassifier

model_rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model_rf.fit(x_train, y_train)

y_pred_rf = model_rf.predict(x_test)
```

```python
rf_accuracy = accuracy_score(y_test, y_pred_rf)

print("Random Forest Accuracy:", rf_accuracy)
```

```python
import pandas as pd

final_results = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        '5-Fold Cross Validation',
        'GridSearchCV Logistic Regression',
        'RandomizedSearchCV Logistic Regression',
        'Random Forest'
    ],

    'Accuracy': [
        baseline_accuracy,
        cv_accuracy,
        grid_accuracy,
        random_accuracy,
        rf_accuracy
    ]
})

final_results
```

The baseline Logistic Regression achieved 99.27% accuracy, while the mean 5-fold cross-validation accuracy was 99.06%, showing consistent model performance across different folds. After hyperparameter tuning, both GridSearchCV and RandomizedSearchCV achieved 99.29% accuracy. The Random Forest ensemble model achieved the highest accuracy of 99.40%. Therefore, among the evaluated approaches, Random Forest was the best-performing model based on accuracy.
